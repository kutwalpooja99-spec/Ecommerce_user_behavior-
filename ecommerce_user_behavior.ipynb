# 7: Check Unique Values

```python
# Unique values in each column
for col in df.columns:
    print("\nColumn:", col)
    print(df[col].unique())
```

```python
print(df.dtypes)

# Convert categorical columns to string
categorical_cols = [
    "gender",
    "device_type",
    "discount_seen",
    "ad_clicked",
    "returning_user",
    "purchase"
]

for col in categorical_cols:
    if col in df.columns:
        df[col] = df[col].astype("string")
```

```python
df = df.dropna(how="all")

print("Shape after removing empty rows:", df.shape)
```

```python
# Numerical columns
num_cols = [
    "age",
    "time_on_site",
    "pages_viewed",
    "previous_purchases",
    "cart_items",
    "avg_session_time",
    "bounce_rate"
]

# Fill numerical missing values with median
for col in num_cols:
    df[col] = df[col].fillna(df[col].median())


# Categorical columns
cat_cols = [
    "gender",
    "device_type",
    "discount_seen",
    "ad_clicked",
    "returning_user"
]

# Fill categorical missing values with mode
for col in cat_cols:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].mode()[0])


# Check remaining null values
print(df.isnull().sum())
```

```python
df.isnull().sum()
```

```python
# Select numerical columns
numeric_columns = df.select_dtypes(
    include="number"
).columns

# Count negative values in each column
negative_values = (df[numeric_columns] < 0).sum()

print("Negative values in each column:")
print(negative_values)
```

```python
print("Both columns missing:",
      (df["user_id"].isnull() & df["purchase"].isnull()).sum())

print("Only user_id missing:",
      (df["user_id"].isnull() & df["purchase"].notnull()).sum())

print("Only purchase missing:",
      (df["user_id"].notnull() & df["purchase"].isnull()).sum())
```

```python
df_cleaned = df.dropna(subset=["user_id", "purchase"]).copy()

print("Before Cleaning:", df.shape)
print("After Cleaning:", df_cleaned.shape)

print(df_cleaned.isnull().sum())
```

# Gender Distribution (Count Plot)

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="gender", hue="gender", palette="Set2", legend=False)

plt.title("Customer Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Number of Customers")
plt.show()
```

```python
# Customer Age Distribution (Histogram)
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(data=df, x="age", bins=20, kde=True, color="skyblue")

plt.title("Customer Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
# Device Type Distribution
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="device_type", hue="device_type",
              palette="viridis", legend=False)

plt.title("Customers by Device Type")
plt.xlabel("Device Type")
plt.ylabel("Number of Customers")
plt.show()
```

# Time Spent on Website vs Purchase

```python
plt.figure(figsize=(9, 5))

sns.boxplot(
    data=df,
    x="purchase",
    y="time_on_site",
    hue="purchase",
    palette="Set2",
    legend=False
)

plt.title("Time Spent on Website vs Purchase")
plt.xlabel("Purchase Status")
plt.ylabel("Time on Site")
plt.show()
```
